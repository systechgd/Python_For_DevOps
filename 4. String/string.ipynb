### 🔢 Indexing

- Indexing means accessing individual characters in a string using their position (index).

- Indexing starts from `0` for the first character.

````python
text = "Hello"
print(text[0])
print(text[1])
print(text[2])
print(text[3])
print(text[4])
text[4] = "T" 
````

````python
text = "Hello"
print(text[-1])
````

### ▶️ Positive Indexing

| Character | H | e | l | l | o |
|:---------:|:-:|:-:|:-:|:-:|:-:|
| Index     | 0 | 1 | 2 | 3 | 4 |



````python
text = "Hello"
print(text[0])  # Output: H
print(text[2])  # Output: l
````

#

### ◀️ Negative Indexing

Each character in a string also has a **negative index** when counting from the end.

| Character | H | e | l | l | o |
|:---------:|:-:|:-:|:-:|:-:|:-:|
| Index     | -5 | -4 | -3 | -2 | -1 |


````python
text = "Hello"
print(text[-1])  # Output: o
print(text[-5])  # Output: H
````

#

### ✂️ Slicing

- Slicing is used to get a substring by specifying a range of indexes.

#### Syntax:

```python
string[start:stop:step]
```

````python
text = "Hello, World!"
print(text[: :])
print(text[0:5])   
print(text[7:])   
print(text[:5])    
print(text[::2])   
print(text[-5:-2])
print(text[: : -1])
print(text)
print(text[15])
````

### ✏️ Modifying Strings Using Built-in Methods

````python
# converting into lower case

text = "hello"
print(text.upper()) 
print(text)
````

````python
# converting into uppercase 

text = "HELLO"
print(text.lower()) 
````

````python
# Remove Leading and Trailing Whitespace
text = "   Hello World   "
print(text.strip())
````

````python
# Remove starting whitespace
text = "  Hello World"
print(text.rstrip())
````

````python
# Remove ending whitespace
text = "Hello       "
print(text.lstrip())
````

````python
# Replace substring

text = "I love Java"
print(text.replace("Java", "Python"))
print(text)
````

````python
# get length of a string

text = "Python"
print(len(text))
````

````python
# Find Index of Substring

text = "Python programming"
print(text.find("gram"))
````

````python
# Capitalize first letter

text = "python is fun"
print(text.capitalize())
````

````python
# Capitalize first of each word

text = "hello world"
print(text.title())
````

````python
# Check Start/End

text = "Hello, World!"
print(text.startswith("Hello")) 
print(text.endswith("!"))        
````

### F-Strings

- F-String was introduced in Python 3.6, and is now the preferred way of formatting strings

````python
price = 59
txt = f"The price is {price} dollars"
print(txt)
````

````python
price = 59
txt = f"The price is {price:.2f} dollars"
print(txt)
````

````python
txt = f"The price is {20 * 59} dollars"
print(txt)
````

## 🔐 Python String Escape Characters – With Examples

| Escape Code | Meaning                    | Code Example                                  | Output                          |
|-------------|----------------------------|-----------------------------------------------|----------------------------------|
| `\\`        | Backslash                  | `print("This is a backslash: \\\\")`          | This is a backslash: \          |
| `\'`        | Single quote               | `print('It\\'s Python')`                      | It's Python                     |
| `\"`        | Double quote               | `print("She said, \\\"Hello!\\\"")`           | She said, "Hello!"              |
| `\n`        | Newline                    | `print("Line1\\nLine2")`                      | Line1<br>Line2                  |
| `\t`        | Tab                        | `print("Name:\\tPython")`                     | Name: Python                    |
| `\r`        | Carriage Return            | `print("12345\\rXY")`                         | XY345                           |
| `\b`        | Backspace                  | `print("abc\\bdef")`                          | abdef                           |
| `\f`        | Form Feed (rare)           | `print("Hello\\fWorld")`                      | HelloWorld                     |
| `\ooo`      | Octal value (e.g., `\141`) | `print("\\141\\142\\143")`                    | abc                             |
| `\xhh`      | Hex value (e.g., `\x61`)   | `print("\\x61\\x62\\x63")`                    | abc                             |

---




````python
print("Backslash: \\")
print("Single quote: \'")
print("Double quote: \"")
print("New line:\nNext line")
print("Tab:\tTabbed")
print("Carriage return: 123\rXY")
print("Backspace: ab \bdef")
print("Form feed: Hello \fWorld")
print("Octal: \141\142\143")  # a b c
print("Hex: \x61\x62\x63")    # a b c
````

### 🧩 Common Parameters:

- The `print()` function in Python has several **optional keyword arguments** that let you control its behavior.

### 📋 Parameters Table

| Parameter | Description                       | Default        |
|-----------|-----------------------------------|----------------|
| `sep`     | Separator between multiple values | `' '` (space)  |
| `end`     | What to print at the end          | `'\n'` (newline) |
| `file`    | Where to send the output          | `sys.stdout`   |
| `flush`   | Whether to flush the output buffer| `False`        |



````python
# sep=

print("Python", "Java", "C++", sep=", ")
````

````python
# end=   By default, Python ends with a newline \n. You can change it to a space, comma, or anything.

print("Hello", end=" ")
print("World")
````
