# ✅ Python Function

- block of reusable code that performs a specific task
- you define it once and call it wherever you need it

## 🧠 Why Use Functions?

- To organize code into chunks
- To reuse code
- To make your code more readable and maintainable
- To avoid repetition

### 🔧 How to Define a Function

```python
def function_name(parameters):
    # Code block (function body)
    return value  # (optional)
```

- `def`: keyword to define a function

- `function_name`: name you give to the function

- `parameters`: optional values you pass to it

- `return: optional`; sends a result back to the caller

```python
# Example 

def greet():
    print("Hello! Welcome to Python.")

# Call the function
greet()
```

### 🧮 Example: Function with Parameters

```python
def add_numbers(a, b):
    result = a + b
    print("Sum:", result)

add_numbers(5, 3)
```

### 🔁 Example: Function with Return Value

```python
def multiply(x, y):
    return x * y

product = multiply(4, 5)
print("Product:", product)
```

### 👥 Example: Function with Default Parameters

```python
def say_hello(name="User"):
    print(f"Hello, {name}!")

say_hello("Alice")
say_hello()
```

### 🌟 Example: Function with Multiple Return Values

```python
def calculate(a, b):
    sum_ = a + b
    diff = a - b
    return sum_, diff

s, d = calculate(10, 5)
print("Sum:", s)
print("Difference:", d)
```

## 🌍 Global vs. Local Scope in Python

#### 📌 Scope refers to the region of the code where a variable is recognized or accessible.

#

### 🏠 Local Scope

- A variable declared inside a function is local to that function.

- It cannot be accessed outside the function.

```python
def my_func():
    message = "Hello from inside!"  # local variable
    print(message)

my_func()

print(message)  #❌ This would raise an error (message not defined here)
```

### 🌐 Global Scope

- A variable declared outside all functions is global.

- It can be accessed anywhere in the code, including inside functions (read-only unless declared global)

```python
name = "Alice"  # global variable

def greet():
    print("Hello", name)

greet()
```

#### ⚠️ Modifying Global Variables Inside a Function
- If you want to modify a global variable inside a function, you need to use the global keyword.
#### ⚠️ Without global (won’t work):

```python
count = 0

def increment():
    count += 1  # ❌ Error: local variable 'count' referenced before assignment

increment()
```

```python
count = 0

def increment():
    global count
    count += 1

increment()
print(count)
```

#

### 🧪 Bonus: nonlocal (for nested functions)
Used to modify a variable from the outer function inside a nested function.

```python
def outer():
    count = 0
    def inner():
        nonlocal count
        count += 1
        print(count)
    inner()

outer()
```

```python
def outer():
    count = 0
    def inner():
        # nonlocal count
        count += 1
        print(count)
    inner()

outer()
```
