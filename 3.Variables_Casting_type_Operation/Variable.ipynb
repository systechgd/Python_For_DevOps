## Variable

- A variable is like a container or label that stores information you want to use later.

```sh
# Syntax of creating variable

variable_name = value
```

````python
user_name = "Anil"

print(user_name)
````

### Rules for Valid Variable Names
- Variable name can only contain `uppercase and lowercase letters`, digits (`0-9`) and underscore (`_`).

- Variable name only `starts with letters or underscore`

- Variables are `Case Sensitive`

### Recomondation for variable naming

- Always provide descriptive name for your variable

````python
## Valid Variables
user_name = "Ganesh"
_username = "Ganesh"
UserName = "Ganesh"
username = "Ganesh"
UserName123 = "Ganesh"
User123_Name = "Ganesh"
````

````python
## Invalid Variables
1Username = "Ganesh"
#username = "Ganesh"
@userName = "Ganesh"
User-Name = "Ganesh"
````

### Assigning multiple value variables

````python
# multiple variables, multiple values
x, y, z = "Orange", "Banana", "Cherry"
print(x)
print(y)
print(z)
````

````python
# one value to multiple variables
x = y = z = "Orange"
print(x)
print(y)
print(z)
````

````python
# output multiple variable value at once
print(x, y, z)
````
