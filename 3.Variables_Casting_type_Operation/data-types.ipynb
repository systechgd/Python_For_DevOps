# 📦 Collection Data Types in Python

| Data Type        | Example                           | Description                          |
|------------------|-----------------------------------|--------------------------------------|
| `list`           | `[1, 2, 3]`                        | Ordered, changeable, allows duplicates |
| `tuple`          | `(1, 2, 3)`                        | Ordered, unchangeable (immutable), Allow Duplicates     |
| `set`            | `{1, 2, 3}`                        | Unordered, no duplicates, Unchangeable              |
| `dict` (dictionary) | `{"name": "Alice", "age": 25}` | Key-value pairs, Duplicate key values will overwrite                       |


#

### Basic Data Types Examples

```python
# To Declare Empty Value
number = 0
pi = 0.0
name = ""
is_active = False

print(number, pi, name, is_active)
print("number : ", type(number),"\n", "pi : ", type(pi), "\n", "name : ", type(name), "\n", "is_active : ", type(is_active))
```

```python
# To declare with data type of value
age = 25
height = 5.6
name = "Alice"
is_student = True

print(age, height, name, is_student)
print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
```

#

### Collection Data Type example

```python
## Declare empty value data type
grades = []
coordinates = ()
unique_numbers = set()
person = {}

print(grades, coordinates, unique_numbers, person)
print("Empty list: ", type(grades))
print("Empty Tuple : ", type(coordinates))
print("Empty Set : ", type(unique_numbers))
print("Empty Dictionary : ", type(person))
```

```python
# with value
grades = [90, 85, 88]
coordinates = (1, 2, 3)
unique_numbers = {1, 2, 3}
person = {"name": "Alice", "age": 25}

print(grades, coordinates, unique_numbers, person)
print("grades: ", type(grades))
print("coordinates : ", type(coordinates))
print("unique_numbers : ", type(unique_numbers))
print("person : ", type(person))
```

### Python Casting
- when you want to specify a type on to a variable

```python
# Define integer type of value
x = int(1)   # x will be 1
y = int(2.8) # y will be 2
z = int("3") # z will be 3

print("x :",type(x), " ", x)
print("y :",type(y), " ", y)
print("z :",type(z), " ", z)
```

```python
# Define float type of variable 
x = float(1)     # x will be 1.0
y = float(2.8)   # y will be 2.8
z = float("3")   # z will be 3.0
w = float("4.2") # w will be 4.2

print("x :", type(x), " ", x)
print("y :", type(y), " ", y)
print("z :", type(z), " ", z)
print("w :", type(w), " ", w)
```

```python
# Define string type of variable
x = str("s1") # x will be 's1'
y = str(2)    # y will be '2'
z = str(3.0)  # z will be '3.0'

print("x :", type(x), " ", x)
print("y :", type(y), " ", y)
print("z :", type(z), " ", z)
```
