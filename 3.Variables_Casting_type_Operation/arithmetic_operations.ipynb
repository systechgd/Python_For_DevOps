## 🧠 PEMDAS Rule (Order of Operations)

**PEMDAS** helps remember the order:
- **P** → Parentheses `()`
- **E** → Exponents `**`
- **M** → Multiplication `*`
- **D** → Division `/`
- **A** → Addition `+`
- **S** → Subtraction `-`

📌 *Multiplication and Division are done left to right, same for Addition and Subtraction.*

---

#

```python
# Examples
# P → Parentheses
result = (2 + 3) * 4    # 5 * 4 = 20
print("Parentheses: ", result)

# E → Exponentiation
result = 2 ** 3         # 8
print("Exponentiation: ", result)

# M & D → Multiplication and Division
result = 10 * 2 / 5     # 20 / 5 = 4.0
print("Multiplication and Division: ", result)

# A & S → Addition and Subtraction
result = 10 + 5 - 3     # 15 - 3 = 12
print("Addition and Subtraction: ", result)

# Combined example:
result = 2 + 3 * 4      # 3 * 4 = 12 → 2 + 12 = 14
print("Combined example: ", result)

result = (2 + 3) * 4    # 5 * 4 = 20 → parentheses changes the outcome
print("parentheses changes the outcome: ", result)
```
